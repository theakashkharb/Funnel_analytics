```python
import duckdb
import kagglehub
import pandas as pd
```

```python
con = duckdb.connect()
```

```python
con.sql("""
CREATE OR REPLACE VIEW data AS
SELECT *
FROM read_parquet(
    'https://huggingface.co/datasets/theakashkharb/ecommerce-funnel-analytics/resolve/main/all_data.parquet'
)
""")
```

```python
con.sql("""
SELECT *
FROM data
LIMIT 10
""").df()
```
